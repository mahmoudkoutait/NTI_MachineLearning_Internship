```python
import pandas as pd
```

```python
df = pd.read_csv('Marketing.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=['gender'], inplace=True)
```

```python
for uni in df['age'].unique():
    print(uni)
```

```python
# Try to convert to numbers; bad values become NaN
df['age'] = pd.to_numeric(df['age'], errors='coerce')
# Drop rows where age could not be converted
df.dropna(subset=['age'], inplace=True)
# Ensure float dtype
df['age'] = df['age'].astype(float)
```

```python
df.shape
```

```python
df['gender'].replace({'M' : 1, 'F' : 0}, inplace= True)
```

```python
df['gender'].value_counts()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters= 2, random_state= 2002)
```

```python
kmeans.fit(df);
```

```python
kmeans.inertia_
```

```python
labels = kmeans.labels_
print(labels[:10])
```

```python
from sklearn.metrics import silhouette_score
silhouette_avg = silhouette_score(df, kmeans.labels_)
print("The average silhouette score is :", silhouette_avg)
```

```python
from sklearn.cluster import DBSCAN
db=DBSCAN(eps=5,min_samples=6,metric='euclidean')
```

```python
model=db.fit(df)
```

```python
label=model.labels_
label
```

```python
from sklearn import metrics
```

```python
y_means = db.fit_predict(df)
```

```python
silhouette_score_dbscan = metrics.silhouette_score(df, y_means)
print("Silhouette Score for DBSCAN: ", silhouette_score_dbscan)
```
