```python
x = final_df.drop('RETURNED_2ND_YR', axis = 1)
y = final_df['RETURNED_2ND_YR']
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
models = {
    'GaussianNB': (GaussianNB(), {'var_smoothing': [1e-9, 1e-8, 1e-7]}),
    'MultinomialNB': (MultinomialNB(), {'alpha': [0.1, 0.5, 1.0]}),
    'BernoulliNB': (BernoulliNB(), {'alpha': [0.1, 0.5, 1.0], 'binarize': [0.0, 0.5, 1.0]})
}
best_models = {}
```

```python
for name, (model, params) in models.items():
    grid = GridSearchCV(model, param_grid=params, cv=5, scoring='accuracy')
    grid.fit(X_train, y_train)
    
    print(f"Best parameters for {name}: {grid.best_params_}")
    print(f"Best CV score: {grid.best_score_:.4f}")
    
    y_pred = grid.predict(X_test)
    print(f"Test set performance for {name}:")
    print(classification_report(y_test, y_pred))
    print("-" * 60)
    
    best_models[name] = grid.best_estimator_
```
